# The `matplotlib` Library

The next topic we are going to cover is data visualization. We will focus a lot on techniques to make "nice" diagrams and what message we want to send to the reader.

Matplotlib is Python's standard plotting library. Virtually any type of diagram you require, Matplotlib has it. Moreover, there are many features you can tweak for your specific plot. Let's cover some today. For a full list of what Matplotlib has to offer, see the documentation here: https://matplotlib.org/stable/api/index.html.

````python
import matplotlib.pyplot as plt 
import numpy as np
````

Given the following data

````python
heartrate = [68,70,71, 71, 
             71, 74, 75, 78, 
             80, 79, 75, 70, 68]
time = ['12:00', '12:05', '12:10', '12:15', 
        '12:20', '12:25', '12:30', '12:35',
        '12:40', '12:45', '12:50', '12:55', '1:00']
````

we can use `plt.scatter()` to plot `time` along the x-axis and `heartrate` along the y-axis.

````python
plt.scatter(time, heartrate)
````

Without adding any extra features (simply using `plt.scatter()`), the plot can be very uninformative. 

Let's add some tweaks:
- We will rotate the x-labels so they aren't horizontal
- Add an x-axis label and a y-axis label
- Add an informative title

We will use the following:
- `plt.xticks(rotation=45)`
- `plt.xlabel('x_axis_name_of_your_choice')` and `plt.ylabel('y_axis_name_of_your_choice')`
- `plt.title('title_of_your_choice')`

To implement these, run all of these <i> in the same cell </i>.

````python
plt.scatter(time, heartrate)
plt.xticks(rotation=45)
plt.xlabel('x_axis_name_of_your_choice') 
plt.ylabel('y_axis_name_of_your_choice')
plt.title('title_of_your_choice')
````

Recall:
- `np.linspace(start, stop, slice_count)` produces `slice_count` amount of numbers between `start` and `stop`
- `np.sin(x)` calculates $\sin(x)$ where `x` is a single number or numpy array (which `np.linspace` produces)
- `np.pi` produces the irrational number $\pi\approx 3.14159$

Using these functions, we can 
1. make a list `xdata` that contains 100 evenly spaced points between 0 and $2\pi$
2. make a list `ysindata` that contains $\sin(x)$ for each $x$ in `xdata`
3. use `plt.plot(xdata, ysindata)` to plot the data

````python
xdata = np.linspace(0, 2*np.pi, 100)
ysindata = np.sin(xdata)
plt.plot(xdata, ysindata)
````

We can us the same `xdata` from above to make a list `ycosdata` that contains $\cos(x)$ for each $x$ in `xdata`.

We can then plot `xdata` against both `ycosdata` and `ysindata` simultaneously. To do so, use `plt.plot(xdata, ysindata)` and `plt.plot(xdata, ycosdata)` in the same cell.

````python
ycosdata = np.cos(xdata)
````

````python
plt.plot(xdata, ysindata) 
plt.plot(xdata, ycosdata)
````

So far, these just look like two curves! To differentiate between the two curves, we can use the `label` attribute and a legend. 

All in one cell:
1. Instead of `plt.plot(xdata, ysindata)`, use `plt.plot(xdata, ysindata, label='Sine')`. Likewise with the cosine data. 
2. Use `plt.legend()` to plot the legend.

````python
plt.plot(xdata, ysindata, label='Sine')
plt.plot(xdata, ycosdata, label='Cosine')
plt.legend()
````

Now we know which curve is which! But, we don't have axis labels or a title. We add these in the following cell.

````python
plt.xlabel('x value')
plt.ylabel('Function value')
plt.title('Cosine vs Sine')
plt.plot(xdata, ysindata, label='Sine')
plt.plot(xdata, ycosdata, label='Cosine')
plt.legend()
````

````python
data = [4, 2, 4, 5, 0, 2, 4, 4, 3, 6, 1, 4, 5, 3, 1, 2, 3, 3, 4, 6, 4, 5,
       2, 6, 5, 2, 3, 3, 4, 4, 2, 1, 4, 3, 2, 3, 6, 7, 3, 3, 3, 4, 2, 4,
       3, 2, 2, 2, 4, 3, 2, 3, 4, 2, 5, 5, 3, 3, 5, 3, 2, 6, 2, 2, 4, 4,
       2, 3, 4, 5, 2, 3, 4, 6, 2, 5, 2, 4, 5, 1, 5, 3, 3, 5, 0, 2, 3, 5,
       3, 2, 2, 5, 7, 5, 3, 3, 4, 2, 4, 4]

# part 1

def count(dataset):
    counts = []
    for i in set(dataset):
        counts.append(dataset.count(i))
    return counts

# part 2-4

labels = [0, 1, 2, 3, 4, 5, 6, 7]
counts = count(data)
plt.pie(counts)
plt.title("This is a Pie Chart")
plt.pie(counts, labels = labels, autopct='%1.1f%%')
plt.show()

# part 5

plt.hist(counts)
plt.title('This is a Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')
````
