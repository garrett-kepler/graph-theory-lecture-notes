Section 5

````python
# 1
def underscore(contlist):
    string = contlist[0]
    for i in contlist[1:]:
        string = string+'_'+i
    return string
underscore(['this', 'is', 'a', 'list'])

# 2
def factorial(n):
    prod = 1
    for i in range(1,n+1):
        prod *= i
    return prod
factorial(3)

# 3
def fibonacci(n):
    fib1 = 0
    fib2 = 1
    for i in range(n):
        fib1, fib2 = fib2, fib1+fib2
    return fib2
for i in range(5):
    print(fibonacci(i))

# 4
def binary(n):
    if n <= 0:
        return "0"
    binarie = ""
    while n > 0:
        remainder = n % 2               
        binarie += str(remainder)
        n //= 2                       
    return binarie[::-1]
binary(18)

# 5
def money(string):
    string = string.replace('$', '')
    string = string.replace(',', '')
    return string
money('$10,000')

# 6 
def reverse(string):
    return string[::-1]
reverse('string')

# 7
def summation(contlist):
    summ = 0
    for i in contlist:
        summ += i
    return summ
summation([1,2,3])

# 8
def minimum(contlist):
    minimum = contlist[0]
    for i in contlist:
        if i<minimum:
            minimum = i
    return minimum
minimum([1,2,3])
````
